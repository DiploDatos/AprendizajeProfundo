## Stop words

```python
nltk.download("stopwords") # Descargamos las stop words
stop_words_english = set(stopwords.words("english"))
print(f"Cantidad de stop words en ingles: {len(stop_words_english)}")
print(f"Algunos ejemplos: {list(stop_words_english)[:20]}")
```

```python
stop_words_spanish = set(stopwords.words("spanish"))
print(f"Cantidad de stop words en español: {len(stop_words_spanish)}")
print(f"Algunos ejemplos: {list(stop_words_spanish)[:20]}")
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words_spanish = set(stopwords.words("english"))
filtered_list_eng = []
for word in words_lst:
  # casefold es una manera de trabajar indistintamente mayusculas y minusculas
  # ya que las stop words están todas en minúscula
   if word.casefold() not in stop_words_english:
        filtered_list_eng.append(word)

print(f"Cantidad de palabras iniciales: {len(words_lst)}")
print(words_lst)
print('\n')
print(f"Cantidad de palabras luego de remover stop words: {len(filtered_list_eng)}")
print(filtered_list_eng)
```

## Lemmatizing

```python
lemmatizer = WordNetLemmatizer()
examples_words = ['discovery'	,'disco'	,'discoveries'	,'discovering']
for word in examples_words:
  print(f"Original word: {word}")
  print(f"Lemmatizer word: {lemmatizer.lemmatize(word)}")
  print('--------')
```

```python
lemmatizer.lemmatize("biggest") #Por defecto toma considera "nouns"
```

El parametro `pos` es por defecto `'n'` donde `n` está asociado a `noun`. Sin embargo, para asegurarnos que `"worst"` sea tratado como un adjetivo  necesitamos cambiar el parámetro por defecto a `pos="a"`.

```python
lemmatizer.lemmatize("biggest", pos="a")
```

```python
import gensim.downloader as api
from gensim.models import FastText

try:
    ft_model = api.load("fasttext-wiki-mini")
except ValueError:
    print("fasttext-wiki-mini not found. Trying to load a larger model (fasttext-wiki-news-subwords-300). This may take a while.")
    try:
        ft_model = api.load("fasttext-wiki-news-subwords-300")
    except Exception as e:
        print(f"Could not load FastText model: {e}")
        ft_model = None

if ft_model:
    print(f'Loaded {len(ft_model.index_to_key)} words.')

    word_vector = ft_model['computer']
    print(f"Vector for 'computer' (first 10 dimensions): {word_vector[:10]}")
    print(f"Dimension of the vector: {len(word_vector)}")

    similar_words = ft_model.most_similar('computer')
    print(f"Words most similar to 'computer': {similar_words}")

    oov_word = 'datascience'
    if oov_word in ft_model:
        oov_vector = ft_model[oov_word]
        print(f"\nVector for OOV word '{oov_word}' (first 10 dimensions): {oov_vector[:10]}")
        print(f"Dimension of the OOV vector: {len(oov_vector)}")
    else:
        print(f"\n'{oov_word}' not found in vocabulary, but FastText can still generate a vector.")
        try:
             oov_vector_fallback = ft_model[oov_word]
             print(f"Vector for OOV word '{oov_word}' (first 10 dimensions): {oov_vector_fallback[:10]}")
             print(f"Dimension of the OOV vector: {len(oov_vector_fallback)}")
        except KeyError:
             print(f"Could not generate vector for '{oov_word}' even with subwords.")
```

```python
from transformers import pipeline

# Cargamos un modelo que soporta embeddings
embedder = pipeline("feature-extraction", model="bert-base-uncased")

text = "The capital of France is Paris."
embedding = embedder(text)

# El resultado es una lista de embeddings por token
print(f"Embedding shape: {len(embedding[0])} x {len(embedding[0][0])}")
```

```python
embedding
```

```python
from transformers import pipeline

unmasker = pipeline('fill-mask', model='bert-base-uncased')

text = "The capital of France is [MASK]."
result = unmasker(text)

for prediction in result:
    print(f"Token: {prediction['token_str']}, Score: {prediction['score']:.4f}")
```
